```python
import numpy as np
```

```python
ax = np.array([1000, 0.2, 0.3, 0.4])
bx = np.array([900, 0.1, 0.15, 0.2])
ay = ax.prod()
by = bx.prod()

pieces = np.log(ax) - np.log(bx)            # step 2: log growth of each factor
total  = np.log(ay) - np.log(by)            # = pieces.sum()
share  = pieces / total                     # step 3: adds up to 1
contrib = share * (ay - by)                 # step 4: adds up to ay - by
rate = (ay - by) / total                    # the "exchange rate" = log mean L(ay, by)

print(pieces, pieces.sum(), total)
print(share, share.sum())
print(contrib, contrib.sum(), ay - by)
print(rate, rate * pieces)                  # same as contrib
```

```python
plain = ax - bx
logs  = np.log(ax) - np.log(bx)

print("plain adds up:", np.isclose(plain.sum(), ay - by))                    # True -> y is a sum
print("log adds up:  ", np.isclose(logs.sum(), np.log(ay) - np.log(by)))     # True -> y is a product
```

```python
ax = np.array([490.0, 165.2, 390.6, 42.0])   # after:  Term Life, Whole Life, Health, Other claims (M)
bx = np.array([432.0, 165.0, 336.0, 40.0])   # before
ay = ax.sum()                                # y is a SUM now
by = bx.sum()

pieces = ax - bx                             # step 2: plain change of each term (no log)
total  = ay - by                             # = pieces.sum()
share  = pieces / total                      # step 3: adds up to 1
contrib = share * (ay - by)                  # step 4: adds up to ay - by
rate = (ay - by) / total                     # "exchange rate" = 1, already in units of y

print(pieces, pieces.sum(), total)
print(share, share.sum())
print(contrib, contrib.sum(), ay - by)
print(rate, rate * pieces)                   # same as contrib
```

```python
plain = ax - bx
logs  = np.log(ax) - np.log(bx)

print("plain adds up:", np.isclose(plain.sum(), ay - by))                    # True -> y is a sum
print("log adds up:  ", np.isclose(logs.sum(), np.log(ay) - np.log(by)))     # True -> y is a product
```
